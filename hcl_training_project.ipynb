```python
import numpy as np
import pandas as pd

data = pd.read_csv("students_dataset_60.csv")
print("Student Records:")
print(data)
print("\nTotal students:", len(data))

avg_marks = np.mean(data["Marks"])
print("Average marks:", round(avg_marks, 2))

query = input("\nEnter Name or Roll No to search: ").strip()
matched = False


for idx in range(len(data)):
    r_no = str(data.loc[idx, "Roll No"]).strip()
    st_name = str(data.loc[idx, "Name"]).strip()
    if r_no.lower() == query.lower() or st_name.lower() == query.lower():
        print("\nRecord found:")
        print(data.iloc[[idx]])
        matched = True
        break

if not matched:
    print("\nRecord not found:")
```
